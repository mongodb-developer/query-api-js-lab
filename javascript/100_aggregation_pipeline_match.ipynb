# Aggregation Pipeline

## What can aggregation stages do?

[Aggregation stages](https://www.mongodb.com/docs/manual/reference/operator/aggregation-pipeline/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) let us process documents step by step inside a pipeline.

In this section, we will use stages such as `$match`, `$project`, `$sort`, and `$limit` to filter documents, select specific fields, order the results, and control how many documents are returned.

## Startup code

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## $match

We’ll use [`$match`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/match/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) to return only the book with the title "Treasure of the Sun".

```javascript
const matchStage = { $match: { title: "Treasure of the Sun" } };
const pipeline = [matchStage];

const result = db.books.aggregate(pipeline);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```

## $project

We’ll [`$project`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/project/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) only the `title` and `year` fields for the book "Treasure of the Sun", excluding the `_id` field from the result.

```javascript
const matchStage = { $match: { title: "Treasure of the Sun" } };

const projectStage = {
    $project: {
        title: 1,
        year: 1,
        _id: 0
    }
};

const pipeline = [matchStage, projectStage];

const result = db.books.aggregate(pipeline);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```

## $limit

We’ll first match books published after 2010, then use [`$limit`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/limit/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) to return only 3 documents.

```javascript
const matchStage = { $match: { year: { $gt: 2010 } } };
const limitStage = { $limit: 3 };

const pipeline = [matchStage, limitStage];

const result = db.books.aggregate(pipeline);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```

## $sort

We’ll [`$sort`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/sort/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) the books by the number of pages in descending order, limit the result to 3 documents, and project only the title and pages fields.

```javascript
const projectStage = {
    $project: {
        title: 1,
        pages: 1,
        _id: 0
    }
};

const sortStage = { $sort: { pages: -1 } };
const limitStage = { $limit: 3 };

const pipeline = [
    projectStage,
    sortStage,
    limitStage
];

const result = db.books.aggregate(pipeline);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```
