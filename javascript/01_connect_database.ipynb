# Connecting to MongoDB

We'll start with some examples to test our connection to MongoDB and how to send commands to the database.

## Select the JavaScript Kernel

To run the code cells in these notebooks you need to select the appropriate kernel (programming language). In this case is JavaScript. Click on the `Select Kernel` button at the top right, then `Jupyter Kernel` > `JavaScript`

The code in these notebooks uses the [MongoDB Shell (`mongosh`)](https://www.mongodb.com/docs/mongodb-shell/) syntax, so everything you learn here can be typed directly in `mongosh`.

## Install the MongoDB Driver

We use npm to install the [MongoDB Node.js driver](https://www.mongodb.com/docs/drivers/node/current/). The dependencies are listed in `package.json`, so you only need to run `npm install` once inside the `javascript` folder (the dev container already does this for you).

The `mongosh-shim.js` file next to the notebooks is a small helper that lets the Node.js driver understand the same syntax as the MongoDB Shell, for example `db.books.find()`.

The Node.js driver is asynchronous, so in these notebooks we write `await` before the calls that talk to the database. That is the only difference with the MongoDB Shell: if you copy the code into `mongosh`, just remove the `await` keywords.

## Set your connection String and connect

We can just connect to `localhost` as this devcontainer is also running a MongoDB instance with some data, or you can connect to your own Atlas cluster just changing the `connectionString` below.

The `connect()` function returns `db`, which points to the `library` database.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Ping the Database server

The `ping` database command fetches information about the state of the MongoDB database.

```javascript
try {
    const response = await db.runCommand({ ping: 1 });
    print("=> Print result of the '{ping: 1}' command.");
    printjson(response);
} catch (e) {
    print(e);
}
```

## Send a Hello command

The `hello` command returns database server details for the backend MongoDB database used in this workshop.

```javascript
try {
    const response = await db.runCommand({ hello: 1 });
    print("=> Print result of the '{hello: 1}' command.");
    printjson(response);
} catch (e) {
    print(e);
}
```

## Get a list of all databases and collections

We'll get a list of all the databases in my cluster using the `listDatabases` command and then we'll print them.

```javascript
print("=> Print list of databases");
const { databases } = await db.adminCommand({ listDatabases: 1 });

for (const { name } of databases) {
    const collectionNames = await db.getSiblingDB(name).getCollectionNames();

    collectionNames.forEach(collectionName =>
        print(name + "." + collectionName)
    );
}
```
