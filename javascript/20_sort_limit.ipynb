# CRUD: Sort & Limit

## What can `sort()` and `limit()` do?

The [`sort()`](https://www.mongodb.com/docs/manual/reference/method/cursor.sort/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) and [`limit()`](https://www.mongodb.com/docs/manual/reference/method/cursor.limit/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) operations let us control the order and number of documents returned by a query.

In this section, we will sort documents in ascending or descending order and limit how many results are returned from the collection.

## Startup code

This cell connects to MongoDB and gives us the `db` variable that points to the `library` database. The `books` collection used throughout the CRUD examples is available as `db.books`.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Sort books by descending number of pages

In this example, we use [`$lte`](https://www.mongodb.com/docs/manual/reference/operator/query/lte/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to filter books published before or in `1930`, project only the `title` and `pages` fields, and use [`sort()`](https://www.mongodb.com/docs/manual/reference/method/cursor.sort/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to order the results by the number of pages in descending order (`-1`).

```javascript
const cursor = db.books
    .find({ year: { $lte: 1930 } }, { title: 1, pages: 1, _id: 0 })
    .sort({ pages: -1 });

await cursor.forEach(printjson);
```

## Retrieve the top 5 longest books

In this example, we retrieve all books, project only the `title` and `pages` fields, sort the results by the number of pages in descending order, and use [`limit()`](https://www.mongodb.com/docs/manual/reference/method/cursor.limit/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to return only the first 5 documents.

```javascript
const cursor = db.books
    .find({}, { title: 1, pages: 1 })
    .sort({ pages: -1 })
    .limit(5);

await cursor.forEach(printjson);
```

## Skip the first 180 books and retrieve the next 5

In this example, we project only the `title` and `pages` fields, sort the books by the number of pages in ascending order (`1`), use [`skip()`](https://www.mongodb.com/docs/manual/reference/method/cursor.skip/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to ignore the first 180 matching documents, and return the next 5 results.

```javascript
const projection = { title: 1, pages: 1 };

const cursor = db.books
    .find({}, projection)
    .sort({ pages: 1 })
    .skip(180)
    .limit(5);

await cursor.forEach(printjson);
```
