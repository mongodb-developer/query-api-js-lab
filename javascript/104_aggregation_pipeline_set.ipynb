# Aggregation Pipeline 

## What does `$set` do?

[$set](https://www.mongodb.com/docs/manual/reference/operator/aggregation/set/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) adds a new field to each document or updates the value of an existing field.

It keeps all the original fields unless another stage removes them later.

## Startup code

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Before $set

In this example, we will first inspect a few documents from the `reviews` collection. Then, we will use `$set` to add a new field called `reviewYear` based on the `timestamp` field.

```javascript
await db.reviews.aggregate([
    { $project: { name: 1, rating: 1, timestamp: 1, bookId: 1, _id: 0 } },
    { $limit: 5 }
]).forEach(printjson);
```

Now we use `$set` to add a new field called `reviewYear`, extracting the year from the `timestamp` field.

```javascript
const setStage = {
    $set: {
        reviewYear: { $year: "$timestamp" }
    }
};

await db.reviews.aggregate([
    setStage,
    { $project: { name: 1, rating: 1, timestamp: 1, reviewYear: 1, bookId: 1, _id: 0 } },
    { $limit: 5 }
]).forEach(printjson);
```

## Summary

With `$set`, we can add a new field to each document or update the value of an existing one while keeping the original fields in the result.
