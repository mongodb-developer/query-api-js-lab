# CRUD: Find and Project

## What can the `find()` operation do?

The [Find documents](https://www.mongodb.com/docs/manual/reference/method/db.collection.find/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) operation lets us query and retrieve documents from a MongoDB collection.

In this section, we will use `find()` with query operators such as `$eq`, `$gt`, `$gte`, and `$and` to search for documents and apply conditions.

We will also use **projections** such as `{ field: 1 }` and `{ _id: 0 }` to return only the fields we need.

## Startup Code

This cell configures the connection to MongoDB, connects to the `library` database, and gives us the `db` variable and the helper functions we will use throughout the notebook. The `books` collection used in the examples is available as `db.books`.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Find one book

This is the **Read** part of CRUD. We use [`find()`](https://www.mongodb.com/docs/manual/reference/method/db.collection.find/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to query documents from a collection.

In this example, [`findOne()`](https://www.mongodb.com/docs/manual/reference/method/db.collection.findOne/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) returns the first matching document from the `books` collection, or `null` if there is none.

```javascript
const firstBook = await db.books.findOne();

if (firstBook) {
    print("Book:");
    printjson(firstBook);
} else {
    print("No book was found.");
}
```

# Find and Project

In this example, we use [`$eq`](https://www.mongodb.com/docs/manual/reference/operator/query/eq/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to find the book with the title `"Treasure of the Sun"` and a [projection](https://www.mongodb.com/docs/manual/tutorial/project-fields-from-query-results/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to return only the `title`, `_id` and `pages` fields.

The first argument of `find()` is the **filter** and the second one is the **projection**.

```javascript
const filter = { title: { $eq: "Treasure of the Sun" } };
const projection = { title: 1, pages: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Find books written after 2021

In this example, we use [`$gt`](https://www.mongodb.com/docs/manual/reference/operator/query/gt/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello), which means **greater than**, to filter books whose `year` is greater than `2021`, and project only the `title`, `_id` and `year`.

```javascript
const filter = { year: { $gt: 2021 } };
const projection = { title: 1, year: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Find all books with 2000 pages or more

We use [`$gte`](https://www.mongodb.com/docs/manual/reference/operator/query/gte/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello), which means **greater than or equal to**, to filter books whose `pages` value is `2000` or higher.

Then, we use a projection to return the `title`, `year`, `genres`, and `_id` fields.

```javascript
const filter = { pages: { $gte: 2000 } };
const projection = { title: 1, year: 1, genres: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Exclude the `_id` Field

In the previous example, MongoDB still returned the `_id` field because it is included by default, even when we do not explicitly list it in the projection.

To remove it, we set `_id: 0` in the [projection](https://www.mongodb.com/docs/manual/tutorial/project-fields-from-query-results/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello). The `_id` field is the only one that can be excluded in a projection that also includes other fields.

```javascript
const filter = { pages: { $gte: 2000 } };
const projection = { title: 1, pages: 1, _id: 0 };

await db.books.find(filter, projection).forEach(printjson);
```
