# Basic CRUD: Update

## What can the `update` operation do?

The [`update`](https://www.mongodb.com/docs/manual/reference/update-methods/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) operations let us modify existing documents in a MongoDB collection.

In this section, we will use operations such as `updateOne()` and `updateMany()` to update fields in documents from the `books` collection.

## Startup code

This cell connects to MongoDB and gives us the `db` variable that points to the `library` database. The `books` collection used in the update examples is available as `db.books`.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Insert one book

In this example, we insert a new document into the `books` collection.
We will later update this same document using `updateOne()`.

```javascript
await db.books.insertOne({
    _id: "quijote",
    title: "El Quijote",
    year: 1501
});
```

### Find inserted book

Before updating the document, we query the `books` collection to confirm that the `El Quijote` document was successfully inserted.

```javascript
await db.books.find({ _id: "quijote" }).forEach(printjson);
```

### Update the inserted book

In this example, we use `updateOne()` to modify the `El Quijote` document.

The update operation:
- adds a new field called `newField` using `$set`
- adds `"Chivalry"` to the `genres` array using `$addToSet`
- updates the `lastUpdated` field with the current timestamp using `$currentDate`

```javascript
const query = { _id: "quijote" };

// Creates instructions to update the values of three document fields
const updates = {
    $set: { newField: 99 },
    $addToSet: { genres: "Chivalry" },
    $currentDate: { lastUpdated: { $type: "timestamp" } }
};

// Instructs the server to insert a new document if none match the query
const options = { upsert: true };

try {
    // Updates the first document that has an "_id" value of "quijote"
    const result = await db.books.updateOne(query, updates, options);
    // Prints the number of updated documents and the upserted document ID, if an upsert was performed
    print("Modified document count: " + result.modifiedCount);
    print("Upserted id: " + result.upsertedId);

// Prints a message if any exceptions occur during the operation
} catch (error) {
    print("Unable to update due to an error: " + error);
}
```

### Verify the updated document

After running the update operation, we query the `books` collection again to confirm that the document was successfully modified.

```javascript
const aBook = await db.books.findOne({ _id: "quijote" });

print("Book:");
printjson(aBook);
```

## Understanding `upsert`

[`upsert`](https://www.mongodb.com/docs/manual/reference/method/db.collection.updateOne/#std-label-updateOne-upsert/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) combines **update** and **insert** behavior in a single operation.

If a matching document exists, MongoDB updates it.
If no document matches the filter, MongoDB inserts a new one instead.

To demonstrate this behavior, we first delete the `El Quijote` document so it no longer exists in the collection.

```javascript
const deleteResult = await db.books.deleteOne({ _id: "quijote" });

// Check the number of deleted documents
const deletedCount = deleteResult.deletedCount;
print("Deleted " + deletedCount + " document(s).");
```

### Run the update operation with `upsert`

Now that the document no longer exists in the collection, we run the update operation again with `upsert: true` enabled.

```javascript
const query = { _id: "quijote" };

// Creates instructions to update the values of three document fields
const updates = {
    $set: { newField: 99 },
    $addToSet: { genres: "Chivalry" },
    $currentDate: { lastUpdated: { $type: "timestamp" } }
};

// Instructs the server to insert a new document if none match the query
const options = { upsert: true };

try {
    // Updates the first document that has an "_id" value of "quijote"
    const result = await db.books.updateOne(query, updates, options);
    // Prints the number of updated documents and the upserted document ID, if an upsert was performed
    print("Modified document count: " + result.modifiedCount);
    print("Upserted id: " + result.upsertedId);

// Prints a message if any exceptions occur during the operation
} catch (error) {
    print("Unable to update due to an error: " + error);
}
```

### Verify the upserted document

Since the document did not exist, MongoDB inserted a new one during the update operation with `upsert: true` enabled.

We can now query the collection again to confirm that the document was created.

```javascript
const upsertedBook = await db.books.findOne({ _id: "quijote" });

print("Book:");
printjson(upsertedBook);
```
