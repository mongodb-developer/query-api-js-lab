# Aggregation Pipeline: $merge
    

## Startup code

````javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
````

### $Merge

 This aggregation filters books with more than 1000 pages and writes the results into a new collection called `books_with_more_than_1000_pages` using [`$merge`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/merge/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello).

````javascript
await db.books.aggregate([
    { $match: { pages: { $gt: 1000 } } },
    { $merge: { into: "books_with_more_than_1000_pages" } }
]).toArray();
````

This prints the books from the new collection, displaying only the title and pages fields.

````javascript
await db.books_with_more_than_1000_pages.aggregate([
    { $project: { _id: 0, title: 1, pages: 1 } }
]).forEach(printjson);
````
