# Basic CRUD: Delete

## What can the `delete` operation do?

The [`delete`](https://www.mongodb.com/docs/manual/reference/delete-methods/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) operations let us remove documents from a MongoDB collection.

In this section, we will use operations such as `deleteOne()` and `deleteMany()` to remove documents from the `books` collection.

## Startup code

This cell connects to MongoDB and gives us the `db` variable that points to the `library` database. The `books` collection used in the delete examples is available as `db.books`.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Insert one book

We want to make sure this book exists in the collection so we can later remove it using a delete operation.

```javascript
await db.books.insertOne({
    _id: "platero",
    title: "Platero y yo",
    year: 1900
});
```

### Find the inserted book

In this example, we query the `books` collection using the `_id` field to confirm that the `Platero y yo` document exists before deleting it.

```javascript
await db.books.find({ _id: "platero" }).forEach(printjson);
```

### Delete one book

In this example, we use `deleteOne()` to remove the document with `_id` equal to `"platero"` from the `books` collection.

The returned result lets us verify how many documents were deleted using `deletedCount`.

```javascript
const deleteResult = await db.books.deleteOne({ _id: "platero" });

// Check the number of deleted documents
const deletedCount = deleteResult.deletedCount;

print("Deleted " + deletedCount + " document(s).");
```
