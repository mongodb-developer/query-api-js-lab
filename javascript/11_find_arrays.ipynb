# CRUD: Querying Array Fields

## How can we query array fields in MongoDB?

MongoDB lets us [query an array](https://www.mongodb.com/docs/manual/tutorial/query-arrays/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) fields in different ways depending on the matching behavior we want.

In this section, we will use operators such as `$all`, `$in`, and `$eq` to find documents where array fields contain specific values, multiple values, or exact array matches.

## Startup code

This cell connects to MongoDB and gives us the `db` variable that points to the `library` database. The `books` collection used in the examples is available as `db.books`.

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## Find books that contain `Poetry`

In this example, we use [`$eq`](https://www.mongodb.com/docs/manual/reference/operator/query/eq/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) (the default when we just write the value) to find books where the `genres` array contains the value `"Poetry"`.

```javascript
const filter = { genres: "Poetry" };
const projection = { title: 1, genres: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Find books that contain both `Family Life` and `Fiction` (AND)

In this example, we use [`$all`](https://www.mongodb.com/docs/manual/reference/operator/query/all/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to find books where the `genres` array contains both `"Family Life"` and `"Fiction"`.

```javascript
const filter = { genres: { $all: ["Family Life", "Fiction"] } };
const projection = { title: 1, genres: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Find books that contain `Family Life` or `Fiction` (OR)

In this example, we use [`$in`](https://www.mongodb.com/docs/manual/reference/operator/query/in/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) to find books where the `genres` array contains either `"Family Life"` or `"Fiction"`.

```javascript
const filter = { genres: { $in: ["Family Life", "Fiction"] } };
const projection = { title: 1, genres: 1 };

await db.books.find(filter, projection).forEach(printjson);
```

## Understanding exact array matches

When using [`$eq`](https://www.mongodb.com/docs/manual/reference/operator/query/eq/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) with an array, MongoDB looks for an **exact array match**.

This means:
- the values must be the same
- the order must also match exactly

For example, the query below returns a document because the `genres` array is exactly `["Poetry", "American"]`.

```javascript
const filter = { genres: ["Poetry", "American"] };
const projection = { title: 1, genres: 1 };

const result = await db.books.findOne(filter, projection);

if (result) {
    print("Book:");
    printjson(result);
} else {
    print("No matching books were found.");
}
```

However, if we invert the values to `["American", "Poetry"]`, the query does not return any documents because the array order no longer matches the stored document exactly.

```javascript
const filter = { genres: ["American", "Poetry"] };
const projection = { title: 1, genres: 1 };

const result = await db.books.findOne(filter, projection);

if (result) {
    print("Book:");
    printjson(result);
} else {
    print("No matching books were found.");
}
```
