# Aggregation Pipeline

## What does `$unwind` do?

[$unwind](https://www.mongodb.com/docs/manual/reference/operator/aggregation/unwind/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) deconstructs an array field and creates one output document for each element in the array.

It keeps the other fields in the document, but repeats them for each generated output document.

## Startup code

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## $unwind

This pipeline first selects the book with `_id` `"0004127382"`, then uses `$unwind` to split the `attributes` array into separate documents, and finally uses `$project` to return only the `title` and `attributes` fields.

```javascript
const result = db.books.aggregate([
    { $match: { _id: "0004127382" } },
    { $unwind: "$attributes" },
    { $project: { title: 1, attributes: 1, _id: 0 } }
]);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```
