# Aggregation Pipeline

## What does `$lookup` do?

[$lookup](https://www.mongodb.com/docs/manual/reference/operator/aggregation/lookup/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) joins documents from another collection and adds the matching results to a new array field.

It keeps the original document and appends the related documents returned by the join.

## Startup code

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## $lookup

This pipeline uses `$lookup` to join data from the `books` collection into each author document. It matches the values in the author's `books` array with the `_id` field in the `books` collection and stores the matching documents in a new field called `booksWritten`.

```javascript
const result = db.authors.aggregate([
    {
        $lookup: {
            from: "books",            // collection with the book data to read into authors
            localField: "books",      // array in authors documents with the ids of all books written by an author
            foreignField: "_id",      // field in the books collection to link with
            as: "booksWritten"        // name of a newly created field with all this info
        }
    },
    {
        $project: {
            name: 1,
            "booksWritten.title": 1,
            "booksWritten.pages": 1,
            "booksWritten.year": 1,
            _id: 0
        }
    },
    { $limit: 20 }
]);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```
