# Basic CRUD: Insert

## What can the `insert` operation do?

The [`insert`](https://www.mongodb.com/docs/manual/reference/insert-methods/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-javascript-workshop&utm_term=ricardo.mello) operations let us add new documents to a MongoDB collection.

In this section, we will use operations such as `insertOne()` and `insertMany()` to create and store documents in the `books` collection.

## Startup code

This cell connects to MongoDB and gives us the `db` variable that points to the `library` database. The `books` collection used in the insert examples is available as `db.books`.

`````javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
`````

## CRUD: Insert

### Insert one book

In this example, we create a new document representing a book and use `insertOne()` to add it to the `books` collection.

The result contains the `insertedId`, the `_id` that MongoDB generated for the new document.

`````javascript
const result = await db.books.insertOne({ title: "El Quijote", year: 1500 });

printjson(result);
`````

### Find the inserted document

In this example, we query the `books` collection to verify that the document inserted with the year `1500` was successfully stored in MongoDB.

`````javascript
await db.books.find({ year: 1500 }).forEach(printjson);
`````

### Find the same document using the `_id`

Can you find the same document again, but this time using the `_id` field instead of the `year`?

Fix the code below by adding the correct `ObjectId`.

`````javascript
const bookFrom1500 = { _id: new ObjectId("") }; // add the _id here!

await db.books.find(bookFrom1500).forEach(printjson);
`````
