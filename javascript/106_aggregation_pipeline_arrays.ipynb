# Aggregation Pipeline

## What does `$match` do with array operators?

[$match](https://www.mongodb.com/docs/manual/reference/operator/aggregation/match/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=aggregation-framework-javascript-workshop&utm_term=ricardo.mello) filters documents based on a condition.

When used with array operators, it can search for documents whose array fields contain specific values. For example, `$all` requires all specified values to be present, while `$in` matches documents that contain at least one of them.

## Startup code

```javascript
// Set your connection String
const connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Load the helper that lets us use the MongoDB Shell (mongosh) syntax in this notebook.
// It provides `print()`, `printjson()` and `ObjectId`, and connects to MongoDB.
const { connect, print, printjson, ObjectId } = require("./mongosh-shim.js");

// Define our database and collection. We'll use the `db` variable that points to the `library` database and `db.books` that points to the collection we're using.
const { db } = await connect(connectionString);
```

## $match - arrays

### $match with the `$all` operator

[arrays-reference](
https://mongodb-developer.github.io/aggregation-pipeline-lab/docs/using-arrays/simple-match-array)

If you want to find books whose `genres` array contains both `"Family Life"` and `"Fiction"`, in any order, use:

```javascript
const result = db.books.aggregate([
    { $match: { genres: { $all: ["Family Life", "Fiction"] } } },
    { $project: { title: 1, genres: 1, _id: 0 } }
]);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```

### $match with the `$in` operator

Use `$in` to find books where the `genres` array contains at least one of the specified values.

```javascript
const result = db.books.aggregate([
    { $match: { genres: { $in: ["Family Life", "Fiction"] } } },
    { $project: { title: 1, genres: 1 } }
]);

// Iterate through the results
await result.forEach(doc => {
    print("book:");
    printjson(doc);
});
```
